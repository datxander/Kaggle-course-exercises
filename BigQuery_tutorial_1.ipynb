```python
from google.cloud import bigquery

# Create an object named "Client"
client = bigquery.Client.from_service_account_info(key_data)
```

```python
# Reference a dataset and then fetch it
dataset_ref = client.dataset("hacker_news", project="bigquery-public-data")
dataset = client.get_dataset(dataset_ref)

print("Successfully connected to the dataset!")
```

```python
# List tables in the dataset
tables = list(client.list_tables(dataset))

print(f"Tables in the dataset '{dataset.dataset_id}':")
for table in tables:
    print(f"- {table.table_id}")
```

```python
# Now reference and fetch a table from this "full" dataset

# Build reference to table
table_ref = dataset_ref.table("full")


# API request to fetch the table
table = client.get_table(table_ref)
```

```python
# Get the table structure through the schema command
table.schema
```

```python
# Pull a sample of 100 records from this table

client.list_rows(table, max_results = 100).to_dataframe()
```
